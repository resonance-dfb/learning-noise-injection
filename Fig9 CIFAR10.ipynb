```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import matplotlib as mpl
import scienceplots

plt.style.use(['science', 'ieee'])
plt.rcParams['font.serif'] = ['Times New Roman']
mpl.rcParams['figure.dpi'] = 800
mpl.rcParams['axes.unicode_minus'] = False

img_paths = [
    'D:/PRE2023 2025/sigmoid.png',
    'D:/PRE2023 2025/laplace.png'
]

# 创建1×2的子图布局
fig, axes = plt.subplots(1, 2, figsize=(6.2, 3))

# 调整子图间距，确保图像不重合
plt.subplots_adjust(left=0.05, right=0.95, top=0.95, bottom=0.05, 
                   wspace=0.1, hspace=0.1)

labels = ['(a)', '(b)']

for i, ax in enumerate(axes):
    img = mpimg.imread(img_paths[i])
    ax.imshow(img, extent=[0, 1, 0, 1])  # 标准化图片范围
    ax.axis('off')
    
    # 设置轴的边距为0
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.margins(0)
    
    # 添加标签到图像右上角
    ax.text(0.0, 1.0, labels[i], transform=ax.transAxes,
            fontsize=14, fontweight='bold', va='top', ha='right')

# 注释掉进一步减少间距的代码，避免图像重合
# for ax in axes:
#     ax.set_position(ax.get_position().shrunk(1.02, 1.02))

# 保存时使用更严格的边距控制
plt.savefig('D:/PRE2025/Fig8.eps', format='eps', dpi=800,
            bbox_inches='tight', pad_inches=0.0, facecolor='white')

plt.show()
```

```python
import matplotlib.pyplot as plt 
import matplotlib as mpl
import scienceplots

plt.style.use(['science', 'ieee'])
plt.rcParams['font.serif'] = ['Times New Roman']
mpl.rcParams['figure.dpi'] = 800
mpl.rcParams['axes.unicode_minus'] = False

sigma_values_2 = [
    0.8174, 0.8168, 0.8035, 0.7945, 0.7777, 0.7771, 0.7850, 0.7777, 0.7696, 0.7647,
    0.7656, 0.7536, 0.7482, 0.7420, 0.7303, 0.7336, 0.7288, 0.7333, 0.7334, 0.7322,
    0.7303, 0.7247, 0.7201, 0.7210, 0.7108, 0.7124, 0.7044, 0.6950, 0.6884, 0.6786,
    0.6741, 0.6631, 0.6587, 0.6497, 0.6372, 0.6359, 0.6213, 0.6066, 0.5932, 0.5870,
    0.5814, 0.5720, 0.5650, 0.5592, 0.5471, 0.5467, 0.5463, 0.5353, 0.5258, 0.5215
]

# 创建epoch编号，从0开始
epochs = list(range(0, len(sigma_values_2)))   # 0 到 49

# 创建图形
plt.figure(figsize=(6, 4))

# 绘制曲线
plt.plot(epochs, sigma_values_2, 
         linewidth=1.5, 
         color='blue', 
         marker='o', 
         markersize=5)

plt.xlabel('Epoch number', fontsize=14)
plt.ylabel(r'$\sigma$', fontsize=14)

# 设置x轴范围和刻度
plt.xlim(0, 50)
plt.xticks(range(0, 51, 5))   # 每5个epoch一个刻度

# 设置y轴范围
plt.ylim(0.4, 1)

# 设置刻度字体大小
plt.tick_params(axis='both', which='major', labelsize=14)
plt.xticks(range(0, 51, 10))
# 设置网格
plt.grid(True, alpha=0.3)

# 调整布局
plt.tight_layout()

# 保存图像
plt.savefig('D:/PRE2025/Fig9.eps', format='eps', dpi=800, bbox_inches='tight')

# 显示图像
plt.show()
```
