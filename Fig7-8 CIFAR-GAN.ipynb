```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, utils
import os
import numpy as np
from torch.utils.data import DataLoader, Dataset
from torch.nn.functional import sigmoid, interpolate
from torchvision.models import inception_v3
from scipy import linalg
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt

# 检查已下载的文件
cache_dir = os.path.expanduser("~/.cache/torch/hub/checkpoints/")
pytorch_fid_file = os.path.join(cache_dir, "pt_inception-2015-12-05-6726825d.pth")
torch_fidelity_file = os.path.join(cache_dir, "weights-inception-2015-12-05-6726825d.pth")

print("检查已下载的模型文件...")
pytorch_fid_exists = os.path.exists(pytorch_fid_file)
torch_fidelity_exists = os.path.exists(torch_fidelity_file)

if pytorch_fid_exists:
    size_mb = os.path.getsize(pytorch_fid_file) / 1024 / 1024
    print(f"pytorch-fid文件存在: {size_mb:.1f} MB")
else:
    print("pytorch-fid文件不存在")

if torch_fidelity_exists:
    size_mb = os.path.getsize(torch_fidelity_file) / 1024 / 1024
    print(f"torch-fidelity文件存在: {size_mb:.1f} MB")
else:
    print("torch-fidelity文件不存在")

# 使用专业的FID和IS计算包 - 直接尝试使用，不预先检查文件完整性
try:
    from pytorch_fid import fid_score
    from pytorch_fid.inception import InceptionV3
    PYTORCH_FID_AVAILABLE = pytorch_fid_exists  # 只要文件存在就尝试使用
    print(f"pytorch-fid 状态: {'可用' if PYTORCH_FID_AVAILABLE else '不可用'}")
except ImportError:
    PYTORCH_FID_AVAILABLE = False
    print("pytorch-fid 包未安装")

try:
    from torchmetrics.image.fid import FrechetInceptionDistance
    from torchmetrics.image.inception import InceptionScore
    TORCHMETRICS_AVAILABLE = torch_fidelity_exists  # 只要文件存在就尝试使用
    print(f"torchmetrics 状态: {'可用' if TORCHMETRICS_AVAILABLE else '不可用'}")
except ImportError:
    TORCHMETRICS_AVAILABLE = False
    print("torchmetrics 包未安装")

# 修改基础路径
BASE_PATH = "D:/PRE2023 2025"

# -------------------------------
# 1. Activation Function: Vsig
# -------------------------------
class Vsig(nn.Module):
    def __init__(self, init_sigma=1.0):
        super(Vsig, self).__init__()
        self.sigma = nn.Parameter(torch.tensor(init_sigma, dtype=torch.float32))

    def forward(self, input):
        eps = 1e-6
        sigma_safe = self.sigma + eps
        output = 1 / (1 + torch.exp(-input / sigma_safe))
        return output

# -------------------------------
# 2. CIFAR-10 Generator & Discriminator
# -------------------------------
class CIFAR10Generator(nn.Module):
    def __init__(self, latent_dim=100, ngf=64):
        super().__init__()
        self.latent_dim = latent_dim
        
        self.main = nn.Sequential(
            nn.ConvTranspose2d(latent_dim, ngf * 8, 4, 1, 0, bias=False),
            nn.BatchNorm2d(ngf * 8),
            nn.ReLU(True),
            nn.ConvTranspose2d(ngf * 8, ngf * 4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf * 4),
            nn.ReLU(True),
            nn.ConvTranspose2d(ngf * 4, ngf * 2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf * 2),
            nn.ReLU(True),
            nn.ConvTranspose2d(ngf * 2, 3, 4, 2, 1, bias=False),
            nn.Tanh()
        )

    def forward(self, z):
        return self.main(z.view(z.size(0), z.size(1), 1, 1))

class CIFAR10Discriminator(nn.Module):
    def __init__(self, use_vsig=False, ndf=64):
        super().__init__()
        self.use_vsig = use_vsig
        
        self.main = nn.Sequential(
            nn.Conv2d(3, ndf, 4, 2, 1, bias=False),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf, ndf * 2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ndf * 2),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf * 2, ndf * 4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ndf * 4),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf * 4, 1, 4, 1, 0, bias=False),
        )
        
        self.activation = Vsig() if use_vsig else nn.Sigmoid()

    def forward(self, x):
        logits = self.main(x)
        return self.activation(logits.view(-1, 1))

# -------------------------------
# 3. Training Function
# -------------------------------
def train_cifar10_gan(epochs=50, use_vsig=False, model_name="sigmoid", batch_size=128):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    generator = CIFAR10Generator().to(device)
    discriminator = CIFAR10Discriminator(use_vsig).to(device)
    
    def weights_init(m):
        classname = m.__class__.__name__
        if classname.find('Conv') != -1:
            nn.init.normal_(m.weight.data, 0.0, 0.02)
        elif classname.find('BatchNorm') != -1:
            nn.init.normal_(m.weight.data, 1.0, 0.02)
            nn.init.constant_(m.bias.data, 0)
    
    generator.apply(weights_init)
    discriminator.apply(weights_init)

    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
    ])

    dataset = datasets.CIFAR10("./data", download=True, train=True, transform=transform)
    dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True, num_workers=2)

    lr = 0.0002
    beta1 = 0.5
    opt_G = optim.Adam(generator.parameters(), lr=lr, betas=(beta1, 0.999))
    opt_D = optim.Adam(discriminator.parameters(), lr=lr, betas=(beta1, 0.999))

    loss_fn = nn.BCELoss()
    
    G_losses = []
    D_losses = []

    print(f"开始训练 {model_name} 模型...")
    
    for epoch in range(epochs):
        epoch_d_loss = 0
        epoch_g_loss = 0
        num_batches = 0
        
        for i, (real_imgs, _) in enumerate(dataloader):
            real_imgs = real_imgs.to(device)
            batch_size_current = real_imgs.size(0)

            real_labels = torch.ones(batch_size_current, 1).to(device) * 0.9
            fake_labels = torch.zeros(batch_size_current, 1).to(device) + 0.1

            # 更新判别器
            discriminator.zero_grad()
            
            real_output = discriminator(real_imgs)
            d_loss_real = loss_fn(real_output, real_labels)
            
            z = torch.randn(batch_size_current, 100).to(device)
            fake_imgs = generator(z)
            fake_output = discriminator(fake_imgs.detach())
            d_loss_fake = loss_fn(fake_output, fake_labels)
            
            d_loss = d_loss_real + d_loss_fake
            d_loss.backward()
            opt_D.step()

            # 更新生成器
            generator.zero_grad()
            fake_output = discriminator(fake_imgs)
            g_loss = loss_fn(fake_output, torch.ones(batch_size_current, 1).to(device))
            g_loss.backward()
            opt_G.step()
            
            epoch_d_loss += d_loss.item()
            epoch_g_loss += g_loss.item()
            num_batches += 1

        avg_d_loss = epoch_d_loss / num_batches
        avg_g_loss = epoch_g_loss / num_batches
        G_losses.append(avg_g_loss)
        D_losses.append(avg_d_loss)
        
        print(f"[{model_name}] Epoch {epoch+1}/{epochs} | D_loss: {avg_d_loss:.4f} | G_loss: {avg_g_loss:.4f}")
        
        if (epoch + 1) % 10 == 0:
            save_sample_images(generator, epoch + 1, model_name)

    # 保存模型
    os.makedirs(BASE_PATH, exist_ok=True)
    torch.save(generator.state_dict(), f'{BASE_PATH}/generator_{model_name}.pth')
    torch.save(discriminator.state_dict(), f'{BASE_PATH}/discriminator_{model_name}.pth')
    
    plot_losses(G_losses, D_losses, model_name)
    
    return generator

# -------------------------------
# 4. 保存样本图像
# -------------------------------
def save_sample_images(generator, epoch, model_name, num_samples=64):
    device = next(generator.parameters()).device
    generator.eval()
    
    os.makedirs(f"{BASE_PATH}/samples_{model_name}/", exist_ok=True)
    
    with torch.no_grad():
        z = torch.randn(num_samples, 100).to(device)
        fake_imgs = generator(z)
        utils.save_image(fake_imgs, f'{BASE_PATH}/samples_{model_name}/epoch_{epoch:03d}.png', 
                        normalize=True, nrow=8)
    
    generator.train()

# -------------------------------
# 5. 保存生成图像用于评估
# -------------------------------
def save_generated_images(generator, model_name="sigmoid", num_images=1000):
    device = next(generator.parameters()).device
    generator.eval()

    save_dir = f"{BASE_PATH}/images_fake_{model_name}/"
    os.makedirs(save_dir, exist_ok=True)

    print(f"正在保存 {num_images} 张生成图像...")
    with torch.no_grad():
        for i in tqdm(range(num_images)):
            z = torch.randn(1, 100).to(device)
            fake_img = generator(z).cpu()
            fake_img = (fake_img + 1) / 2
            fake_img = torch.clamp(fake_img, 0, 1)
            img_path = os.path.join(save_dir, f"{i:04d}.png")
            utils.save_image(fake_img, img_path)

# -------------------------------
# 6. 保存真实CIFAR-10图像
# -------------------------------
def save_real_cifar10_images(num_images=1000):
    transform = transforms.Compose([transforms.ToTensor()])
    dataset = datasets.CIFAR10("./data", train=True, download=True, transform=transform)
    
    save_dir = f"{BASE_PATH}/images_real_cifar10/"
    os.makedirs(save_dir, exist_ok=True)
    
    print(f"正在保存 {num_images} 张真实CIFAR-10图像...")
    for i in tqdm(range(num_images)):
        img, _ = dataset[i]
        img_path = os.path.join(save_dir, f"{i:04d}.png")
        utils.save_image(img, img_path)

# -------------------------------
# 7. 使用专业包计算FID和IS
# -------------------------------
def calculate_fid_pytorch_fid(real_path, fake_path, device='cuda'):
    """使用pytorch-fid包计算FID"""
    if not PYTORCH_FID_AVAILABLE:
        print("pytorch-fid 不可用，跳过计算")
        return None
    
    try:
        print("使用pytorch-fid计算FID...")
        fid_value = fid_score.calculate_fid_given_paths([real_path, fake_path], 
                                                       batch_size=50, 
                                                       device=device, 
                                                       dims=2048)
        return fid_value
    except Exception as e:
        print(f"pytorch-fid计算失败: {e}")
        return None

def calculate_metrics_torchmetrics(real_path, fake_path, device='cuda'):
    """使用torchmetrics计算FID和IS"""
    if not TORCHMETRICS_AVAILABLE:
        print("torchmetrics 不可用，跳过计算")
        return None, None, None

    try:
        # 数据预处理
        transform = transforms.Compose([
            transforms.Resize((32, 32)),
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])
        
        real_dataset = ImageFolderDataset(real_path, transform=transform)
        fake_dataset = ImageFolderDataset(fake_path, transform=transform)
        
        real_loader = DataLoader(real_dataset, batch_size=32, shuffle=False)
        fake_loader = DataLoader(fake_dataset, batch_size=32, shuffle=False)
        
        # 初始化指标
        fid = FrechetInceptionDistance(feature=2048, normalize=True).to(device)
        inception_score = InceptionScore(normalize=True).to(device)
        
        print("使用torchmetrics计算指标...")
        
        # 更新FID
        for real_batch in tqdm(real_loader, desc="处理真实图像"):
            real_batch = real_batch.to(device)
            # 确保图像是RGB格式并且在正确范围内
            if real_batch.shape[1] == 1:
                real_batch = real_batch.repeat(1, 3, 1, 1)
            real_batch = (real_batch + 1) / 2  # 转换到[0,1]
            real_batch = torch.clamp(real_batch, 0, 1)
            fid.update(real_batch, real=True)
        
        for fake_batch in tqdm(fake_loader, desc="处理生成图像"):
            fake_batch = fake_batch.to(device)
            if fake_batch.shape[1] == 1:
                fake_batch = fake_batch.repeat(1, 3, 1, 1)
            fake_batch = (fake_batch + 1) / 2
            fake_batch = torch.clamp(fake_batch, 0, 1)
            fid.update(fake_batch, real=False)
            
            # 更新IS
            inception_score.update(fake_batch)
        
        # 计算最终分数
        fid_score = fid.compute().item()
        is_mean, is_std = inception_score.compute()
        is_mean, is_std = is_mean.item(), is_std.item()
        
        return fid_score, is_mean, is_std
        
    except Exception as e:
        print(f"torchmetrics计算失败: {e}")
        return None, None, None

# -------------------------------
# 8. 图像数据集类
# -------------------------------
class ImageFolderDataset(Dataset):
    def __init__(self, root, transform=None):
        self.root = root
        self.transform = transform
        self.images = [f for f in os.listdir(root) if f.endswith(('.png', '.jpg', '.jpeg'))]
        
    def __len__(self):
        return len(self.images)
    
    def __getitem__(self, idx):
        img_path = os.path.join(self.root, self.images[idx])
        image = Image.open(img_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        return image

# -------------------------------
# 9. 自定义IS和FID实现（备用）
# -------------------------------
def compute_inception_score_custom(images, device='cuda', batch_size=32, splits=10):
    """自定义IS计算（当专业包不可用时使用）"""
    inception_model = inception_v3(pretrained=True, transform_input=False).to(device)
    inception_model.eval()
    
    up = nn.Upsample(size=(299, 299), mode='bilinear', align_corners=False)
    
    def get_pred(x):
        if x.shape[1] == 1:
            x = x.repeat(1, 3, 1, 1)
        x = up(x)
        x = inception_model(x)
        return torch.nn.functional.softmax(x, dim=1).data.cpu().numpy()

    dataloader = DataLoader(images, batch_size=batch_size, shuffle=False)
    preds = []
    
    print("计算Inception Score (自定义实现)...")
    with torch.no_grad():
        for batch in tqdm(dataloader):
            batch = batch.to(device)
            pred = get_pred(batch)
            preds.append(pred)
    
    preds = np.concatenate(preds, axis=0)
    
    scores = []
    N = preds.shape[0]
    for i in range(splits):
        part = preds[i * (N // splits): (i + 1) * (N // splits), :]
        py = np.mean(part, axis=0)
        scores.append([np.sum(p * (np.log(p + 1e-10) - np.log(py + 1e-10))) for p in part])
    
    scores = np.concatenate(scores)
    is_mean = np.exp(np.mean(scores))
    is_std = np.exp(np.std(scores))
    
    return is_mean, is_std

def calculate_frechet_distance(mu1, sigma1, mu2, sigma2, eps=1e-6):
    """计算Fréchet距离"""
    mu1 = np.atleast_1d(mu1)
    mu2 = np.atleast_1d(mu2)
    
    sigma1 = np.atleast_2d(sigma1)
    sigma2 = np.atleast_2d(sigma2)
    
    diff = mu1 - mu2
    
    covmean, _ = linalg.sqrtm(sigma1.dot(sigma2), disp=False)
    if not np.isfinite(covmean).all():
        offset = np.eye(sigma1.shape[0]) * eps
        covmean = linalg.sqrtm((sigma1 + offset).dot(sigma2 + offset))
    
    if np.iscomplexobj(covmean):
        if not np.allclose(np.diagonal(covmean).imag, 0, atol=1e-3):
            m = np.max(np.abs(covmean.imag))
            raise ValueError('Imaginary component {}'.format(m))
        covmean = covmean.real
    
    tr_covmean = np.trace(covmean)
    
    return (diff.dot(diff) + np.trace(sigma1) + 
            np.trace(sigma2) - 2 * tr_covmean)

def compute_fid_score_custom(real_images, fake_images, device='cuda', batch_size=32):
    """自定义FID计算"""
    inception_model = inception_v3(pretrained=True, transform_input=False).to(device)
    inception_model.eval()
    
    inception_model.fc = nn.Identity()
    
    up = nn.Upsample(size=(299, 299), mode='bilinear', align_corners=False)
    
    def get_activations(dataset):
        dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=False)
        activations = []
        
        with torch.no_grad():
            for batch in tqdm(dataloader):
                batch = batch.to(device)
                if batch.shape[1] == 1:
                    batch = batch.repeat(1, 3, 1, 1)
                batch = up(batch)
                pred = inception_model(batch).cpu().numpy()
                activations.append(pred)
        
        return np.concatenate(activations, axis=0)
    
    print("计算真实图像的特征 (自定义实现)...")
    real_activations = get_activations(real_images)
    print("计算生成图像的特征 (自定义实现)...")
    fake_activations = get_activations(fake_images)
    
    mu_real = np.mean(real_activations, axis=0)
    sigma_real = np.cov(real_activations, rowvar=False)
    
    mu_fake = np.mean(fake_activations, axis=0)
    sigma_fake = np.cov(fake_activations, rowvar=False)
    
    fid_score = calculate_frechet_distance(mu_real, sigma_real, mu_fake, sigma_fake)
    
    return fid_score

# -------------------------------
# 10. 综合评估函数
# -------------------------------
def evaluate_model(fake_dir, real_dir, model_name, device='cuda'):
    """综合评估模型"""
    print(f"\n{'='*50}")
    print(f"评估 {model_name} 模型")
    print(f"{'='*50}")
    
    results = {}
    
    # 1. 尝试使用pytorch-fid计算FID
    fid_pytorch = calculate_fid_pytorch_fid(real_dir, fake_dir, device)
    if fid_pytorch is not None:
        results['FID_pytorch'] = fid_pytorch
        print(f"FID (pytorch-fid): {fid_pytorch:.4f}")
    
    # 2. 尝试使用torchmetrics计算FID和IS
    fid_tm, is_mean_tm, is_std_tm = calculate_metrics_torchmetrics(real_dir, fake_dir, device)
    if fid_tm is not None:
        results['FID_torchmetrics'] = fid_tm
        results['IS_mean_torchmetrics'] = is_mean_tm
        results['IS_std_torchmetrics'] = is_std_tm
        print(f"FID (torchmetrics): {fid_tm:.4f}")
        print(f"IS (torchmetrics): {is_mean_tm:.4f} ± {is_std_tm:.4f}")
    
    # 3. 如果专业包都不可用，使用自定义实现
    if not results:
        print("使用自定义实现计算指标...")
        transform = transforms.Compose([
            transforms.Resize((32, 32)),
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])
        
        fake_dataset = ImageFolderDataset(fake_dir, transform=transform)
        real_dataset = ImageFolderDataset(real_dir, transform=transform)
        
        # 计算IS
        is_mean_custom, is_std_custom = compute_inception_score_custom(fake_dataset, device=device)
        results['IS_mean_custom'] = is_mean_custom
        results['IS_std_custom'] = is_std_custom
        
        # 计算FID
        fid_custom = compute_fid_score_custom(real_dataset, fake_dataset, device=device)
        results['FID_custom'] = fid_custom
        
        print(f"IS (自定义): {is_mean_custom:.4f} ± {is_std_custom:.4f}")
        print(f"FID (自定义): {fid_custom:.4f}")
    
    return results

# -------------------------------
# 11. 绘制损失曲线
# -------------------------------
def plot_losses(G_losses, D_losses, model_name):
    plt.figure(figsize=(10, 5))
    plt.plot(G_losses, label='Generator Loss')
    plt.plot(D_losses, label='Discriminator Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title(f'Training Losses - {model_name}')
    plt.legend()
    plt.grid(True)
    plt.savefig(f'{BASE_PATH}/losses_{model_name}.png')
    plt.show()

# -------------------------------
# 12. 执行训练和评估
# -------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"使用设备: {device}")
print(f"基础路径: {BASE_PATH}")

# 创建基础目录
os.makedirs(BASE_PATH, exist_ok=True)

# 保存真实CIFAR-10图像
print("保存真实CIFAR-10图像...")
save_real_cifar10_images(num_images=1000)

# 训练两个模型
print("\n训练使用Sigmoid激活函数的GAN...")
generator_sigmoid = train_cifar10_gan(epochs=50, use_vsig=False, model_name="sigmoid")

print("\n训练使用Vsig激活函数的GAN...")  
generator_vsig = train_cifar10_gan(epochs=50, use_vsig=True, model_name="vsig")

# 生成图像用于评估
print("\n生成图像用于评估...")
save_generated_images(generator_sigmoid, model_name="sigmoid", num_images=1000)
save_generated_images(generator_vsig, model_name="vsig", num_images=1000)

# 评估模型
results_sigmoid = evaluate_model(
    f"{BASE_PATH}/images_fake_sigmoid/", 
    f"{BASE_PATH}/images_real_cifar10/", 
    "Sigmoid", device=device)

results_vsig = evaluate_model(
    f"{BASE_PATH}/images_fake_vsig/", 
    f"{BASE_PATH}/images_real_cifar10/", 
    "Vsig", device=device)

# 输出最终对比结果
print("\n" + "="*60)
print("最终结果对比:")
print("="*60)

for metric in results_sigmoid.keys():
    if metric in results_vsig:
        print(f"{metric}:")
        print(f"  Sigmoid: {results_sigmoid[metric]:.4f}")
        print(f"  Vsig:    {results_vsig[metric]:.4f}")
        print()
```

```python
# -------------------------------
# 13. 生成对比图像
# -------------------------------
def generate_comparison_images(generator_sigmoid, generator_vsig, num_samples=64, save_individual=True):
    """生成对比图像"""
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    
    # 确保生成器在evaluation模式
    generator_sigmoid.eval()
    generator_vsig.eval()
    
    print(f"\n生成 {num_samples} 张对比图像...")
    
    with torch.no_grad():
        # 使用相同的噪声向量确保公平比较
        z = torch.randn(num_samples, 100).to(device)
        
        # 生成图像
        fake_imgs_sigmoid = generator_sigmoid(z)
        fake_imgs_vsig = generator_vsig(z)
        
        # 归一化到[0,1]
        fake_imgs_sigmoid = (fake_imgs_sigmoid + 1) / 2
        fake_imgs_vsig = (fake_imgs_vsig + 1) / 2
        
        # 确保在正确范围内
        fake_imgs_sigmoid = torch.clamp(fake_imgs_sigmoid, 0, 1)
        fake_imgs_vsig = torch.clamp(fake_imgs_vsig, 0, 1)
        
        # 保存对比网格图
        print("保存Sigmoid生成的图像网格...")
        utils.save_image(fake_imgs_sigmoid, f'{BASE_PATH}/comparison_sigmoid_grid.png', 
                        normalize=False, nrow=8)
        
        print("保存Vsig生成的图像网格...")
        utils.save_image(fake_imgs_vsig, f'{BASE_PATH}/comparison_vsig_grid.png', 
                        normalize=False, nrow=8)
        
        # 创建并排对比图
        print("创建并排对比图...")
        create_side_by_side_comparison(fake_imgs_sigmoid, fake_imgs_vsig, num_display=16)
        
        # 保存单独的对比样本
        if save_individual:
            save_individual_comparisons(fake_imgs_sigmoid, fake_imgs_vsig, num_pairs=10)
    
    # 恢复训练模式
    generator_sigmoid.train()
    generator_vsig.train()

def create_side_by_side_comparison(imgs_sigmoid, imgs_vsig, num_display=16):
    """创建并排对比图"""
    import matplotlib.pyplot as plt
    
    # 选择前num_display张图像
    imgs_sigmoid = imgs_sigmoid[:num_display]
    imgs_vsig = imgs_vsig[:num_display]
    
    # 重新设计布局：每行显示8对图像，共2行
    rows = 2
    cols = 16  # 8对图像，每对占2列
    
    fig, axes = plt.subplots(rows, cols, figsize=(32, 8))
    fig.suptitle('Generated Images Comparison: (a) Sigmoid, (b) Vsig', fontsize=20, y=0.95)
    
    for i in range(min(num_display, 16)):  # 最多显示16张
        row = i // 8  # 每行8对
        col = (i % 8) * 2  # 每对占2列
        
        if row >= rows or col >= cols-1:  # 防止越界
            break
            
        # Sigmoid图像 (左边) - 标记为(a)
        ax_sigmoid = axes[row, col]
        img_sigmoid = imgs_sigmoid[i].cpu().permute(1, 2, 0).numpy()
        ax_sigmoid.imshow(img_sigmoid)
        ax_sigmoid.set_title(f'(a)', fontsize=12, fontweight='bold')
        ax_sigmoid.axis('off')
        
        # Vsig图像 (右边) - 标记为(b)
        ax_vsig = axes[row, col + 1]
        img_vsig = imgs_vsig[i].cpu().permute(1, 2, 0).numpy()
        ax_vsig.imshow(img_vsig)
        ax_vsig.set_title(f'(b)', fontsize=12, fontweight='bold')
        ax_vsig.axis('off')
    
    # 隐藏未使用的子图
    for row in range(rows):
        for col in range(cols):
            pair_idx = row * 8 + col // 2
            if pair_idx >= num_display or col >= (min(num_display, 8) * 2):
                axes[row, col].axis('off')
    
    plt.tight_layout()
    plt.savefig(f'{BASE_PATH}/side_by_side_comparison.png', dpi=300, bbox_inches='tight')
    plt.show()
    print(f"并排对比图已保存: {BASE_PATH}/side_by_side_comparison.png")

def save_individual_comparisons(imgs_sigmoid, imgs_vsig, num_pairs=10):
    """保存单独的对比样本"""
    comparison_dir = f"{BASE_PATH}/individual_comparisons/"
    os.makedirs(comparison_dir, exist_ok=True)
    
    print(f"保存 {num_pairs} 对单独对比图像...")
    
    for i in range(min(num_pairs, len(imgs_sigmoid))):
        # 创建单个对比图
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))
        
        # Sigmoid图像 - 标记为(a)
        img_sigmoid = imgs_sigmoid[i].cpu().permute(1, 2, 0).numpy()
        ax1.imshow(img_sigmoid)
        ax1.set_title('(a) Sigmoid', fontsize=16, fontweight='bold')
        ax1.axis('off')
        
        # Vsig图像 - 标记为(b)
        img_vsig = imgs_vsig[i].cpu().permute(1, 2, 0).numpy()
        ax2.imshow(img_vsig)
        ax2.set_title('(b) Vsig', fontsize=16, fontweight='bold')
        ax2.axis('off')
        
        plt.tight_layout()
        plt.savefig(f'{comparison_dir}/pair_{i+1:02d}.png', dpi=200, bbox_inches='tight')
        plt.close()  # 关闭图像以节省内存
    
    print(f"单独对比图像已保存到: {comparison_dir}")

def create_quality_comparison_grid():
    """创建质量对比网格，展示不同质量的样本"""
    print("\n创建质量对比展示...")
    
    # 读取已保存的图像
    sigmoid_dir = f"{BASE_PATH}/images_fake_sigmoid/"
    vsig_dir = f"{BASE_PATH}/images_fake_vsig/"
    
    if not os.path.exists(sigmoid_dir) or not os.path.exists(vsig_dir):
        print("找不到生成的图像目录，跳过质量对比")
        return
    
    # 随机选择一些图像进行展示
    sigmoid_files = [f for f in os.listdir(sigmoid_dir) if f.endswith('.png')][:20]
    vsig_files = [f for f in os.listdir(vsig_dir) if f.endswith('.png')][:20]
    
    fig, axes = plt.subplots(4, 10, figsize=(25, 10))
    fig.suptitle('Quality Comparison: Random Samples (Top 2 rows: Sigmoid, Bottom 2 rows: Vsig)', fontsize=16)
    
    # 显示Sigmoid样本 (前两行)
    for i in range(20):
        row = i // 10
        col = i % 10
        
        img_path = os.path.join(sigmoid_dir, sigmoid_files[i])
        img = plt.imread(img_path)
        axes[row, col].imshow(img)
        axes[row, col].axis('off')
        if row == 0:
            axes[row, col].set_title(f'S{i+1}', fontsize=8)
    
    # 显示Vsig样本 (后两行)
    for i in range(20):
        row = i // 10 + 2
        col = i % 10
        
        img_path = os.path.join(vsig_dir, vsig_files[i])
        img = plt.imread(img_path)
        axes[row, col].imshow(img)
        axes[row, col].axis('off')
        if row == 2:
            axes[row, col].set_title(f'V{i+1}', fontsize=8)
    
    plt.tight_layout()
    plt.savefig(f'{BASE_PATH}/quality_comparison_grid.png', dpi=300, bbox_inches='tight')
    plt.show()
    print(f"质量对比网格已保存: {BASE_PATH}/quality_comparison_grid.png")

def analyze_image_statistics():
    """分析生成图像的统计特性"""
    print("\n分析生成图像的统计特性...")
    
    # 加载一些生成的图像
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
    ])
    
    sigmoid_dir = f"{BASE_PATH}/images_fake_sigmoid/"
    vsig_dir = f"{BASE_PATH}/images_fake_vsig/"
    
    if not os.path.exists(sigmoid_dir) or not os.path.exists(vsig_dir):
        print("找不到生成的图像目录，跳过统计分析")
        return
    
    # 计算图像统计信息
    def calculate_image_stats(img_dir, num_samples=100):
        files = [f for f in os.listdir(img_dir) if f.endswith('.png')][:num_samples]
        
        pixel_values = []
        for filename in files:
            img_path = os.path.join(img_dir, filename)
            img = Image.open(img_path).convert('RGB')
            img_tensor = transforms.ToTensor()(img)
            pixel_values.extend(img_tensor.flatten().tolist())
        
        pixel_values = np.array(pixel_values)
        return {
            'mean': np.mean(pixel_values),
            'std': np.std(pixel_values),
            'min': np.min(pixel_values),
            'max': np.max(pixel_values)
        }
    
    sigmoid_stats = calculate_image_stats(sigmoid_dir)
    vsig_stats = calculate_image_stats(vsig_dir)
    
    print("\n图像统计对比:")
    print("="*40)
    print(f"{'Metric':<10} {'Sigmoid':<12} {'Vsig':<12}")
    print("-"*40)
    for key in sigmoid_stats.keys():
        print(f"{key:<10} {sigmoid_stats[key]:<12.4f} {vsig_stats[key]:<12.4f}")

# -------------------------------
# 14. 执行图像对比分析
# -------------------------------
print("\n" + "="*60)
print("开始图像对比分析")
print("="*60)

# 生成对比图像
generate_comparison_images(generator_sigmoid, generator_vsig, num_samples=64)

# 创建质量对比网格
create_quality_comparison_grid()

# 分析图像统计特性
analyze_image_statistics()

print("\n" + "="*60)
print("图像对比分析完成！")
print("="*60)
print("生成的对比文件:")
print(f"1. {BASE_PATH}/comparison_sigmoid_grid.png - Sigmoid生成的图像网格")
print(f"2. {BASE_PATH}/comparison_vsig_grid.png - Vsig生成的图像网格")
print(f"3. {BASE_PATH}/side_by_side_comparison.png - 并排对比图")
print(f"4. {BASE_PATH}/quality_comparison_grid.png - 质量对比网格")
print(f"5. {BASE_PATH}/individual_comparisons/ - 单独对比图像文件夹")
print("="*60)
```
